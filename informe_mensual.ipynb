```python
# 1. Cargar datos del CSV
import pandas as pd

df = pd.read_csv('ventas_tienda.csv')

print(df.head())
```

```python
# 2. Calcular ventas totales por mes
df['mes'] = pd.to_datetime(df['fecha']).dt.to_period('M')

ventas_por_mes = df.groupby('mes').apply(lambda d: (d['cantidad'] * d['precio']).sum())
ventas_por_mes = ventas_por_mes.sort_index()
print("Ventas por mes:")
print(ventas_por_mes)
```

```python
# 2. Calcular ventas totales por mes
df['mes'] = pd.to_datetime(df['fecha']).dt.to_period('M')

ventas_por_mes = df.groupby('mes').apply(lambda x: x['cantidad'] * x['precio'].sum())
ventas_por_mes = ventas_por_mes.sort_index()
print("Ventas por mes:")
print(ventas_por_mes)
```

```python
# 3. Determinar producto más vendido vs con mayor ingresos
ventas_por_producto = df.groupby('producto')['cantidad'].sum()
print("Ventas por producto:")
print(ventas_por_producto.sort_values(ascending=False))
producto_mas_vendido = ventas_por_producto.sort_values(ascending=False).idxmax()
print("El producto más vendido es:", producto_mas_vendido)

# 4. Graficar ventas por mes
ventas_por_mes = df.groupby('mes').apply(lambda d: (d['cantidad'] * d['precio']).sum())
ventas_por_mes = ventas_por_mes.sort_index()
print("Ventas por mes:")
print(ventas_por_mes)
```
